## task 1
*Understand how many edges would be necessary to connect together (no matter how faint) every distinct word in the English language.*


The largest binomial coefficient for a given input would at the midpoint for the sequence that the pascals triangle lists and thus is:
$$(\frac{n!}{r!(n-r)!})$$

Where $r$ is the floor of $\frac{n}{2}$.

```python
import numpy as np
from scipy.special import factorial as fac# not commonly used module I guess?

n = 190
k = np.floor(n/2)
edges = fac(n) / (fac(k) * fac(n - k))
print(edges)
```
